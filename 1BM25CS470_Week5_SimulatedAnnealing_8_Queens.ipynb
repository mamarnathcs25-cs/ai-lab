```python
import random
import math

def heuristic(board):
    n = len(board)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or \
               abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts

def simulated_annealing_search(n, initial_board=None):
    if initial_board is not None:
        board = list(initial_board)
    else:
        board = [random.randint(0, n - 1) for _ in range(n)]

    temperature = 100.0
    cooling_rate = 0.99

    while temperature > 0.01:
        current_h = heuristic(board)

        if current_h == 0:
            return board

        col = random.randint(0, n - 1)
        row = random.randint(0, n - 1)

        neighbor = board[:]
        neighbor[col] = row

        delta = heuristic(neighbor) - current_h

        if delta < 0 or random.random() < math.exp(-delta / temperature):
            board = neighbor

        temperature *= cooling_rate

    return board if heuristic(board) == 0 else None

def simulated_annealing(n, initial_board=None):
    while True:
        solution = simulated_annealing_search(n, initial_board)
        if solution is not None:
            return solution
        initial_board = None

def display(board):
    if board is None:
        print("Solution not found")
        return

    for row in range(len(board)):
        print(" ".join(
            "Q" if board[col] == row else "."
            for col in range(len(board))
        ))

n = 8
solution = simulated_annealing(n)
display(solution)
```
