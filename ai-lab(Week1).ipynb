```python

board = ["" for _ in range(9)]


def display():
    print()
    print(" ", board[0], "|", board[1], "|", board[2])
    print("---+---+---")
    print(" ", board[3], "|", board[4], "|", board[5])
    print("---+---+---")
    print(" ", board[6], "|", board[7], "|", board[8])
    print()


def check_winner(player):
    winning_pattern = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_pattern:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


def board_full():
    return "" not in board


def computer_move():
    for i in range(9):
        if board[i] == "":
            board[i] = "O"

            if check_winner("O"):
                return

            board[i] = ""
    for i in range(9):
        if board[i] == "":
            board[i] = "X"

            if check_winner("X"):
                board[i] = "O"
                return

            board[i] = ""


    for i in range(9):
        if board[i] == "":
            board[i] = "O"
            return


print("TIC-TAC-TOE")
print("Human = X")
print("Computer = O")

while True:

    display()


    while True:
        try:
            pos = int(input("Enter the position (1-9): "))

            if pos < 1 or pos > 9:
                print("Enter a number between 1 and 9.")

            elif board[pos - 1] != "":
                print("Position already occupied.")

            else:
                board[pos - 1] = "X"
                break

        except ValueError:
            print("Please enter a valid number.")


    if check_winner("X"):
        display()
        print("Human wins!")
        break

    if board_full():
        display()
        print("Game Draw!")
        break


    computer_move()

    print("Computer made its move.")


    if check_winner("O"):
        display()
        print("Computer wins!")
        break



    if board_full():
        display()
        print("Game Draw!")
        break
```

```python


room_A = input("Enter status of Room A (Dirty/Clean): ").capitalize()
room_B = input("Enter status of Room B (Dirty/Clean): ").capitalize()

location = input("Enter vacuum starting location (A/B): ").upper()

rooms = {
    'A': room_A,
    'B': room_B
}

print("\nInitial State:")
print(rooms)
print("Vacuum Location:", location)

while True:

    if rooms[location] == "Dirty":

        print("Action: SUCK")
        rooms[location] = "Clean"

    else:

        print("Action: MOVE")

        if location == 'A':
            location = 'B'
        else:
            location = 'A'


    if rooms['A'] == "Clean" and rooms['B'] == "Clean":
        break

print("\nFinal State:")
print(rooms)
print("Vacuum Location:", location)
print("Both rooms are clean.")
```
