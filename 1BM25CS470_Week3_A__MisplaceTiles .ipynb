```python
import heapq

# Goal state
GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)


# Heuristic: Number of misplaced tiles
def heuristic(state):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != GOAL[i]:
            count += 1
    return count


# Generate neighboring states
def get_neighbors(state):
    neighbors = []
    blank = state.index(0)
    row, col = divmod(blank, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_blank = r * 3 + c
            new_state = list(state)

            new_state[blank], new_state[new_blank] = (
                new_state[new_blank], new_state[blank]
            )

            neighbors.append(tuple(new_state))

    return neighbors


# A* search algorithm
def astar(start):
    pq = [(heuristic(start), 0, start)]
    g_cost = {start: 0}
    parent = {start: None}

    while pq:
        f, g, current = heapq.heappop(pq)

        if g != g_cost.get(current):
            continue

        if current == GOAL:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            return path[::-1]

        for neighbor in get_neighbors(current):
            new_g = g + 1

            if new_g < g_cost.get(neighbor, float("inf")):
                g_cost[neighbor] = new_g
                parent[neighbor] = current

                new_f = new_g + heuristic(neighbor)
                heapq.heappush(pq, (new_f, new_g, neighbor))

    return None


# Display puzzle
def display(state):
    for i in range(0, 9, 3):
        print(*("_" if x == 0 else x for x in state[i:i+3]))
    print()


# Main program
start = tuple(map(int, input(
    "Enter 9 numbers (0 represents blank): "
).split()))

if len(start) != 9 or set(start) != set(range(9)):
    print("Invalid input!")
else:
    solution = astar(start)

    if solution:
        print("\nSolution found!")
        print("Number of moves:", len(solution) - 1)

        for step, state in enumerate(solution):
            print("Step", step)
            display(state)
    else:
        print("No solution exists.")
```
