```python
def find_blank(state):
    return state.index(0)


def generate_successors(state):
    successors = []
    idx = find_blank(state)
    row, col = divmod(idx, 3)

    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1),
    ]

    for dr, dc in moves:
        new_row, new_col = row + dr, col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_idx = new_row * 3 + new_col

            state_list = list(state)
            state_list[idx], state_list[new_idx] = (
                state_list[new_idx],
                state_list[idx],
            )
            successors.append(tuple(state_list))

    return successors


def dfs(initial_state, goal_state):

    stack = []

    stack.append(initial_state)

    visited = set()

    while stack:
        state = stack.pop()
        if state == goal_state:
            return "SUCCESS"
        if state in visited:
            continue
        visited.add(state)
        successors = generate_successors(state)
        for successor in successors:
            if successor not in visited:
                stack.append(successor)

    return "FAILURE"

if __name__ == "__main__":
    initial = (1, 2, 3, 4, 0, 5, 6, 7, 8)
    goal = (1, 2, 3, 4, 5, 0, 6, 7, 8)

    result = dfs(initial, goal)
    print(f"Result: {result}")
```

```python

def dls(state, goal, depth, limit, path):
    path.append(state)

    if state == goal:
        return True

    if depth == limit:
        path.pop()
        return False
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                x, y = i, j

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dx, dy in moves:
        nx, ny = x + dx, y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            next_state = [row[:] for row in state]

            next_state[x][y], next_state[nx][ny] = (
                next_state[nx][ny], next_state[x][y]
            )

            key = tuple(map(tuple, next_state))
            path_states = [tuple(map(tuple, s)) for s in path]

            if key not in path_states:
                if dls(next_state, goal, depth + 1, limit, path):
                    return True

    path.pop()
    return False


def ids(initial, goal, max_limit=50):
    limit = 0

    while limit <= max_limit:
        path = []

        if dls(initial, goal, 0, limit, path):
            return path

        limit += 1

    return None


initial = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

path = ids(initial, goal)

if path:
    for state in path:
        for row in state:
            print(row)
        print()
else:
    print("Solution not found within depth limit")
```
