```python
def unify(x, y, substitution={}):

    if substitution is None:
        return None

    if x == y:
        return substitution

    if isinstance(x, str) and x.islower():
        substitution[x] = y
        return substitution

    if isinstance(y, str) and y.islower():
        substitution[y] = x
        return substitution

    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            substitution = unify(a, b, substitution)

            if substitution is None:
                return None

        return substitution

    return None

expr1 = ("Eats", "x", "Apple")
expr2 = ("Eats", "Riya", "y")

result = unify(expr1, expr2, {})

if result:
    print("Unification Successful")
    print("Substitution:", result)
else:
    print("Unification Failed")
```

```python

def unify(x, y, substitution=None):

    if substitution is None:
        substitution = {}

    if x == y:
        return substitution

    if isinstance(x, str) and x.islower():
        if x in substitution:
            return unify(substitution[x], y, substitution)
        substitution[x] = y
        return substitution

    if isinstance(y, str) and y.islower():
        if y in substitution:
            return unify(x, substitution[y], substitution)
        substitution[y] = x
        return substitution

    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            substitution = unify(a, b, substitution)

            if substitution is None:
                return None

        return substitution

    return None


# Function to get expression from user
def get_expression(n):
    predicate = input(f"Enter predicate name for expression {n}: ")
    arguments = input("Enter arguments separated by spaces: ").split()

    return tuple([predicate] + arguments)


# Main program
expr1 = get_expression(1)
expr2 = get_expression(2)

result = unify(expr1, expr2, {})

if result is not None:
    print("\nUnification Successful")
    print("Substitution:", result)
else:
    print("\nUnification Failed")
```
